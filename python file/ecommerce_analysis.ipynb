# FINDING OF PROBLEM PRODUCTS

````python
product_analysis = (df.groupby("Product Name")
.agg( Sales = ("Sales","sum"),
     Profit = ("Profit", "sum")
     ))
````

````python
sales_threshold = product_analysis["Sales"].quantile(0.75)
profit_threshold = product_analysis["Profit"].median()

problem_products = product_analysis[
    (product_analysis["Sales"] >= sales_threshold) &
    (product_analysis["Profit"] <= profit_threshold)
]

problem_products.sort_values(
    "Sales",
    ascending=False
).head(20)
````

````python
# here i can say that-
#"Some high-revenue products are not necessarily high-profit products."
````
